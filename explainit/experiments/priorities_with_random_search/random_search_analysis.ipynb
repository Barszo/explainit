# Investigating the Scoring Discrepancy

Let's check what's happening with the gradient-based method for sample 1→2

```python
# Load the data files
import pandas as pd
import numpy as np
import ast

# Load preference results to get exemplars
df_pref = pd.read_csv('experiment_results_preference_based.csv')

# Load standard methods results
df_std = pd.read_csv('experiment_results_standard_methods.csv')

# Safe parsing function
def safe_parse_list(s):
    if pd.isna(s) or s == '' or s == '[]':
        return None
    try:
        return ast.literal_eval(s)
    except:
        # Try to extract numbers using regex
        import re
        numbers = re.findall(r'-?\d+\.?\d*e?[+-]?\d*', str(s))
        if numbers:
            return [float(n) for n in numbers]
        return None

# Parse list columns
for col in ['sample_values', 'exemplar_values', 'cf_values']:
    if col in df_pref.columns:
        df_pref[col] = df_pref[col].apply(safe_parse_list)
    if col in df_std.columns:
        df_std[col] = df_std[col].apply(safe_parse_list)

print("Data loaded successfully")
print(f"Preference results: {len(df_pref)} rows")
print(f"Standard methods results: {len(df_std)} rows")
```

```python
# Get the gradient-based CF for sample 1→2
gradie_1_2 = df_std[(df_std['sample_idx'] == 1) & 
                     (df_std['target_idx'] == 2) & 
                     (df_std['method'] == 'gradient_based')].iloc[0]

print("Gradient-based CF for Sample 1→2:")
print(f"Valid: {gradie_1_2['valid']}")
print(f"Sample values: {gradie_1_2['sample_values']}")
print(f"CF values: {gradie_1_2['cf_values']}")
print(f"\nScore from table:")
print(f"  NoPen: 3.00")
print(f"  Strict: 0.00")
print(f"  PerFeat: 3.00")

# Get exemplar
exemplar_1_2 = df_pref[(df_pref['sample_idx'] == 1) & 
                       (df_pref['target_idx'] == 2)].iloc[0]['exemplar_values']
print(f"\nExemplar values: {exemplar_1_2}")
```

```python
# Now manually calculate the scores to see what's happening
import sys
sys.path.append('/Users/bartosz/projects/explainit/explainit_project')

from explainit.experiments.priorities_with_random_search.experiment_final import AutoMPGExperiment
import pickle

# Load training data
with open('/Users/bartosz/projects/explainit/explainit_project/ML_models_for_tests/ML_regression_results/Auto_MPG/scaler.pkl', 'rb') as f:
    scaler = pickle.load(f)

X_data = pd.read_csv('/Users/bartosz/projects/explainit/explainit_project/ML_models_for_tests/ML_regression_results/Auto_MPG/X_data.csv')
from sklearn.model_selection import train_test_split
X_train, X_test, _, _ = train_test_split(X_data, X_data, test_size=0.2, random_state=42)
X_train_scaled = scaler.transform(X_train).astype(np.float32)

# Create experiment instance
experiment = AutoMPGExperiment()

# Get values
sample = gradie_1_2['sample_values']
exemplar = exemplar_1_2
cf = gradie_1_2['cf_values']
exemplar_weight = 0.01

print("Defining preferences...")
preferences = experiment.define_preferences(sample, exemplar, X_train_scaled, exemplar_weight)
```

```python
# Check each feature
print("\nFeature-by-feature analysis:")
print("="*80)
total_no_penalty = 0.0
total_per_feature = 0.0
violations = {}

for idx in range(4):
    cf_value = cf[idx]
    pref_info = preferences['numerical'][idx]
    pref_func = pref_info['function']
    acceptable_min = pref_info['min']
    acceptable_max = pref_info['max']
    
    # Check if violated
    violated = cf_value < acceptable_min or cf_value > acceptable_max
    if violated:
        if cf_value < acceptable_min:
            violations[idx] = ('below', acceptable_min - cf_value)
        else:
            violations[idx] = ('above', cf_value - acceptable_max)
    
    # Calculate scores
    pref_score = pref_func(cf_value)
    
    print(f"\nFeature {idx} ({experiment.feature_names[idx]}):")
    print(f"  Sample: {sample[idx]:.4f}")
    print(f"  Exemplar: {exemplar[idx]:.4f}")
    print(f"  CF: {cf_value:.4f}")
    print(f"  Acceptable range: [{acceptable_min:.4f}, {acceptable_max:.4f}]")
    print(f"  Violated: {violated}")
    if violated:
        print(f"    Violation: {violations[idx][0]}, distance: {violations[idx][1]:.4f}")
    print(f"  Preference score (no penalty): {pref_score:.4f}")
    
    # Add to totals
    total_no_penalty += pref_score
    if violated:
        print(f"  Per-feature score: 0.0000 (violated)")
        total_per_feature += 0.0
    else:
        print(f"  Per-feature score: {pref_score:.4f}")
        total_per_feature += pref_score

print("\n" + "="*80)
print(f"TOTAL No Penalty: {total_no_penalty:.4f}")
print(f"TOTAL Per-Feature: {total_per_feature:.4f}")
print(f"TOTAL Strict: {0.0 if violations else total_no_penalty:.4f}")
```
