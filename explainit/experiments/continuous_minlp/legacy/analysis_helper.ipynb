# Dataset Analysis Helper

This notebook allows you to explore and analyze categorical and numerical features from the diabetes dataset.

```python
import pandas as pd
import ipywidgets as widgets
from IPython.display import display, clear_output
import os
```

## Load Datasets

```python
# Define paths to the CSV files
base_path = "data_analysis/diabetes/"
categorical_path = os.path.join(base_path, "categorical_features.csv")
numerical_path = os.path.join(base_path, "numerical_features.csv")
model_path = "model_analysis/model_analysis.csv"

# Load datasets
categorical_df = pd.read_csv(categorical_path)
numerical_df = pd.read_csv(numerical_path)
model_df = pd.read_csv(model_path)

print("Datasets loaded successfully!")
print(f"Categorical features shape: {categorical_df.shape}")
print(f"Numerical features shape: {numerical_df.shape}")
```

## Dataset Selection and Display

```python
# Create widgets for dataset selection
dataset_dropdown = widgets.Dropdown(
    options={'Categorical Features': 'categorical', 'Numerical Features': 'numerical'},
    value='categorical',
    description='Dataset:',
    style={'description_width': 'initial'}
)

rows_slider = widgets.IntSlider(
    value=10,
    min=1,
    max=50,
    step=1,
    description='Display Rows:',
    style={'description_width': 'initial'}
)

output_area = widgets.Output()

def display_dataset(dataset_type, num_rows):
    with output_area:
        clear_output(wait=True)
        
        if dataset_type == 'categorical':
            df = categorical_df
            title = "Categorical Features"
        else:
            df = numerical_df
            title = "Numerical Features"
        
        print(f"\n{'='*60}")
        print(f"{title}")
        print(f"{'='*60}")
        print(f"\nShape: {df.shape}")
        print(f"Columns: {list(df.columns)}")
        print(f"\nFirst {num_rows} rows:")
        print(df.head(num_rows).to_string())
        print(f"\nData Info:")
        print(df.info())

# Create interactive widget
interactive_display = widgets.interactive_output(
    display_dataset,
    {'dataset_type': dataset_dropdown, 'num_rows': rows_slider}
)

# Display controls
controls = widgets.VBox([dataset_dropdown, rows_slider])
display(controls)
display(output_area)

# Initial display
display_dataset(dataset_dropdown.value, rows_slider.value)
```

## Quick Statistics

```python
print("\n" + "="*60)
print("Categorical Features Summary")
print("="*60)
print(categorical_df.describe(include='all').to_string())

print("\n" + "="*60)
print("Numerical Features Summary")
print("="*60)
print(numerical_df.describe().to_string())
```

## Feature-by-Feature Analysis

```python
# Create feature selector
all_features = list(categorical_df['feature_name']) + list(numerical_df['feature_name'])

feature_dropdown = widgets.Dropdown(
    options=all_features,
    value=all_features[0] if all_features else None,
    description='Select Feature:',
    style={'description_width': 'initial'}
)

feature_output = widgets.Output()

def display_feature_details(feature_name):
    with feature_output:
        clear_output(wait=True)
        
        # Check if feature is in categorical or numerical
        if feature_name in categorical_df['feature_name'].values:
            feature_data = categorical_df[categorical_df['feature_name'] == feature_name]
            print(f"\nCategorical Feature: {feature_name}")
        else:
            feature_data = numerical_df[numerical_df['feature_name'] == feature_name]
            print(f"\nNumerical Feature: {feature_name}")
        
        print(f"{'-'*60}")
        print(feature_data.to_string(index=False))

feature_dropdown.observe(
    lambda change: display_feature_details(change['new']),
    names='value'
)

display(feature_dropdown)
display(feature_output)

# Initial display
display_feature_details(feature_dropdown.value)
```
